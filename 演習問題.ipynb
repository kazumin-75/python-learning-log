#### 問６．for文とif文を使って、list_bから3の倍数のみを抽出したlist_cを作成してください。

```python
#回答欄
list_c=[]
for i in list_b:
    if i%3==0:
        list_c.append(i)
print (list_b)
```

#### 問７．list_cとenumerateを使って、  
　　　0番目の要素は6です。  
　　　1番目の要素は12です。  
　　　2番目の要素は18です。  
#### 　　　とprintしてください。

```python
#回答欄
for i,j in enumerate(list_c):
    print(str(i)+"番目の要素は"+str(j)+"です")
```
